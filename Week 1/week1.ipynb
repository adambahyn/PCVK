```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
imageurls =["https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg",
"https://raw.githubusercontent.com/opencv/opencv/master/samples/data/butterfly.jpg",
"https://raw.githubusercontent.com/opencv/opencv/master/samples/data/starry_night.jpg",]

for url in imageurls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tingi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
image_3 = image_2.copy()
row = int(image_3.shape[0]/2)

for x in range(image_3.shape[1]):
  image_3[row, x] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
img_task = image_2.copy()
h, w = img_task.shape[:2]
img_task[20:70, 20:70] = [255, 255, 255]

cv2_imshow(img_task)
```

```python
img_loop_2 = image_2.copy()
h, w = img_loop_2.shape[:2]

mid_w = w // 2
for y in range(h):
    img_loop_2[y, mid_w] = [0, 255, 0]

for x in range(w):
    y = int(x * (h - 1) / (w - 1))
    img_loop_2[y, x] = [0, 255, 0]

for x in range(w):
    y = int(x * (h - 1) / (w - 1))
    img_loop_2[h - 1 - y, x] = [0, 255, 0]
cv2_imshow(img_loop_2)
```

```python
img_loop_3 = image_2.copy()
h, w = img_loop_3.shape[:2]
mid_h = h // 2
start_x, end_x = w // 4, 3 * w // 4

for x in range(start_x, end_x):
    img_loop_3[mid_h, x] = [0, 0, 255]

cv2_imshow(img_loop_3)
```
